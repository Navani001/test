```python
import matplotlib.pyplot as plt
import numpy as np

xpoints = np.array([0, 6])
ypoints = np.array([0, 250])

plt.plot(xpoints, ypoints)
plt.show()
```

```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5]
sales = [100, 120, 90, 150, 180]

fig, ax = plt.subplots()

ax.plot(days, sales)

ax.set_title("Sales Over Days")
ax.set_xlabel("Day")
ax.set_ylabel("Sales")

plt.show()
```

```python
import matplotlib.pyplot as plt

height = [160, 165, 170, 175, 180]
weight = [55, 60, 68, 75, 82]

fig, ax = plt.subplots()

ax.scatter(height, weight)

ax.set_title("Height vs Weight")
ax.set_xlabel("Height (cm)")
ax.set_ylabel("Weight (kg)")

plt.show()
```

```python
import matplotlib.pyplot as plt

subjects = ["Python", "NumPy", "Pandas", "Matplotlib"]
marks = [90, 85, 95, 80]

fig, ax = plt.subplots()

ax.bar(subjects, marks)

ax.set_title("Marks by Subject")
ax.set_xlabel("Subject")
ax.set_ylabel("Marks")

plt.show()
```

```python
import matplotlib.pyplot as plt

subjects = ["Python", "NumPy", "Pandas", "Matplotlib"]
marks = [90, 85, 95, 80]

fig, ax = plt.subplots()

ax.barh(subjects, marks)

ax.set_title("Marks by Subject")
ax.set_xlabel("Marks")
ax.set_ylabel("Subject")

plt.show()
```

```python
import matplotlib.pyplot as plt

ages = [
    18, 18, 19, 19, 19,
    20, 20, 20, 20,
    21, 21, 21,
    22, 22,
    23, 23,
    24,
    25,
    28, 30
]

fig, ax = plt.subplots()

ax.hist(ages, bins=11)

ax.set_title("Age Distribution")
ax.set_xlabel("Age")
ax.set_ylabel("Number of Students")

plt.show()
```

```python
import matplotlib.pyplot as plt

marks = [
    60, 65, 68, 70, 72,
    73, 75, 76, 78, 80,
    82, 85, 90, 95, 150
]

fig, ax = plt.subplots()

ax.boxplot(marks)

ax.set_title("Marks Distribution")
ax.set_ylabel("Marks")

plt.show()
```

```python
import matplotlib.pyplot as plt

categories = ["Rent", "Food", "Transport", "Other"]
amounts = [10000, 5000, 3000, 2000]

fig, ax = plt.subplots()

ax.pie(
    amounts,
    labels=categories,
    autopct="%1.1f%%"
)

ax.set_title("Monthly Expense Distribution")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

matrix = np.array([
    [1, 2, 3,10],
    [4, 5, 6,20],
    [7, 100, 9,30],
    [7, 100, 9,30],
    [7, 100, 9,30],
])

fig, ax = plt.subplots()

ax.imshow(matrix)

plt.show()
```

```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5]

temperature = [30, 32, 31, 35, 34]

error = [1, 2, 1, 2, 1]

fig, ax = plt.subplots()

ax.errorbar(
    days,
    temperature,
    yerr=error
)

ax.set_title("Temperature Measurements")
ax.set_xlabel("Day")
ax.set_ylabel("Temperature")

plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

fig, ax = plt.subplots()

ax.plot(x, y)

ax.fill_between(x, y)

plt.show()
```
